```python
text = "PyThOn"
upper_char = []
lower_char = []
upper_count = 0
lower_count = 0 
for i in text :
    if i.isupper():
        upper_char.append(i)
        upper_count += 1
    elif i.islower():
        lower_char.append(i)
        lower_count += 1
print(f"upper_count : {upper_count , upper_char} and lower_count : {lower_count , lower_char}")
```

```python
s = "abcabcdabb"

longest = ""

for i in range(len(s)):
    temp = ""
    for j in range(i, len(s)):
        if s[j] not in temp:
            temp += s[j]
        else:
            break

    if len(temp) > len(longest):
        longest = temp

print("Longest substring:", longest)
print("Length:", len(longest))
```

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
d = {}
for i in words:
    print(i)
    k = "".join(sorted(words))
    if k in d:
        d[k].append(words)
    else:
        d[k] = [words]
    print(k , "and" , d)
    
for j in d.values():
    print(j)
```

```python
words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:
    key = "".join(sorted(word))

    if key in groups:
        groups[key].append(word)
    else:
        groups[key] = [word]

for value in groups.values():
    print(value)
```
