Reverse a string without using [::-1]

Input:  "hello"
Output: "olleh"

```python
input = "hello"
empty = ""
for i in input:
    empty = i + empty
    
print(empty)
```

Check whether a string is a palindrome

Input:  "madam"
Output: True

```python
input = "madam"
rev = input[::-1]
palaindrome = True
if input == rev :
    print(palaindrome)
else :
    palaindrome = False
    print(palaindrome)
    
```

Find the largest and smallest number in a list

Input:  [10, 5, 20, 8, 15]
Output: Largest = 20, Smallest = 5

```python
list =[10, 5, 30, 8, 15]
largest = list[0]
smallest = list[0]
for i in list:
    if i > largest:
        largest = i
    if i < smallest and i != 0 :
        smallest = i 
        
print(largest)
print(smallest)
```

Find duplicate elements in a list

Input:  [1, 2, 3, 2, 4, 1]
Output: [2, 1]

```python
original =  [1, 2, 3, 2, 4, 1]
seen = []
duplicate = []

for i in original:
    if i in seen :
        duplicate.append(i)
    else :
        seen.append(i)
        
print(duplicate)
print(seen)
```

Count the frequency of each character in a string

Input:  "banana"
Output: {'b': 1, 'a': 3, 'n': 2}

```python
input = "banana"
freq = {}

for i in input:
    freq[i] = freq.get(i , 0) +1
    
print(freq)
```

Find the second-largest number in a list

Input:  [10, 5, 20, 8, 15]
Output: 15

```python
Input = [10, 5, 20, 8, 15]

largest = 0 
second_largest = 0
for i in Input:
    if i > largest :
        second_largest = largest 
        largest = i
    elif i > second_largest and i != largest:
        second_largest = i
        
print(largest)
print(second_largest)
```
