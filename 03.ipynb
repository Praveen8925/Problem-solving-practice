```python
# Reverse a string without using [::-1]

# Input:  "hello"
# Output: "olleh"

input = "hello"
empty = ""
for i in input:
    empty = i + empty
    
print(empty)
```

```python
# Check whether a string is a palindrome

# Input:  "madam"
# Output: True

input = "madam"
rev = input[::-1]
palaindrome = True
if input == rev :
    print(palaindrome)
else :
    palaindrome = False
    print(palaindrome)
    
```

```python
# Find the largest and smallest number in a list

# Input:  [10, 5, 20, 8, 15]
# Output: Largest = 20, Smallest = 5

list =[10, 5, 30, 8, 15]
largest = list[0]
smallest = list[0]
for i in list:
    if i > largest:
        largest = i
    if i < smallest and i != 0 :
        smallest = i 
        
print(largest)
print(smallest)
```

```python
# Find duplicate elements in a list

# Input:  [1, 2, 3, 2, 4, 1]
# Output: [2, 1]


original =  [1, 2, 3, 2, 4, 1]
seen = []
duplicate = []

for i in original:
    if i in seen :
        duplicate.append(i)
    else :
        seen.append(i)
        
print(duplicate)
print(seen)
```

```python
# Count the frequency of each character in a string

# Input:  "banana"
# Output: {'b': 1, 'a': 3, 'n': 2}

input = "banana"
freq = {}

for i in input:
    freq[i] = freq.get(i , 0) +1
    
print(freq)
```

```python
# Check whether a number is prime

# Input:  17
# Output: True

input = 17
prime = True 

if input < 2:
    prime = False
else :
    for i in range(2,10):
        if input % i == 0 :
            prime = False
            break
if prime :
    print("prime")
    
else :
    print("not prime")

     
    
```

```python
# Find the factorial of a number

# Input:  5
# Output: 120


input = 5 
f = 1
for i in range(1,input+1):
    f=  f * i
    print(f)
```

```python
# Print the Fibonacci series

# Input:  7
# Output: 0 1 1 2 3 5 8


Input = 7
a = 0
b = 1
for i in range(Input):
    print(a , end=" ")
    c = a + b
    a = b 
    b = c
    
    
```

```python
# Find the second-largest number in a list

# Input:  [10, 5, 20, 8, 15]
# Output: 15


Input = [10, 5, 20, 8, 15]

largest = 0 
second_largest = 0
for i in Input:
    if i > largest :
        second_largest = largest 
        largest = i
    elif i > second_largest and i != largest:
        second_largest = i
        
print(largest)
print(second_largest)
```
